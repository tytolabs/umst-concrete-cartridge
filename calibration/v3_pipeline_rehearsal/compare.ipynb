# V3 pipeline rehearsal (not validation)

Regenerates `tyto_mortar_pipeline_rehearsal.png` from **synthetic** `placeholder_points.json`.

**V3-gate: CLOSED** — pending Term 3 cast. S1 measured τ₀ band only.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

data = json.loads(Path('placeholder_points.json').read_text())
assert data.get('WARNING_not_measurements')
pts = data['points']
ages = [p['age_hours'] for p in pts]
ph = [p['strength_mpa_placeholder'] for p in pts]
pp = [p['strength_mpa_predicted_placeholder'] for p in pts]

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(ages, ph, 'o-', label='synthetic placeholder')
ax.plot(ages, pp, 's--', label='synthetic predicted placeholder')
ax.set_title('tyto_mortar pipeline rehearsal (placeholder data)\nV3-gate: CLOSED')
ax.legend(fontsize=8)
fig.savefig('tyto_mortar_pipeline_rehearsal.png', dpi=150)
```
